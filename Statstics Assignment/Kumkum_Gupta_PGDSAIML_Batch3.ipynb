## Task 1. For the suburb of Altona, it is postulated that a typical property sells for $800,000.
• Use the data at hand to test this assumption.
• Is the typical property price really $800,000 or has it increased?
• Use a significance level of 5%.


```python

from scipy import stats

# filter for Altona
altona_prices = df[df['Suburb'] == 'Altona']['Price'].dropna()

# 2. Descriptive Statistics
mean_price = altona_prices.mean()
n_samples = len(altona_prices)
std_error = stats.sem(altona_prices)

print(f"--- Statistics for Altona ---")
print(f"Number of properties: {n_samples}")
print(f"Mean Price: ${mean_price:}")
print(f"Standard Error: {std_error:,.2f}")
```

```python
# Perform One-Sample T-Test


# Hypothesis:
# Null Hypothesis (H0): Average Price = $800,000
# Alternative Hypothesis (H1): Average Price > $800,000 (Right-tailed test)
# Significance Level (alpha): 0.05

# Population mean under H0
pop_mean_h0 = 800000

# stats.ttest_1samp performs a two-tailed test by default. 
# We specify alternative='greater' for the right-tailed test (has it increased?)
t_stat, p_value = stats.ttest_1samp(altona_prices, pop_mean_h0, alternative='greater')

print(f"\n--- Hypothesis Test Results ---")
print(f"T-Statistic: {t_stat:.4f}")
print(f"P-Value: {p_value:.4f}")


alpha = 0.05
if p_value < alpha:
    print(f"\nConclusion: Since P-Value ({p_value:.4f}) < Alpha ({alpha}), we REJECT the null hypothesis.")
    print("There is sufficient evidence to suggest that the typical property price in Altona is significantly GREATER than $800,000.")
else:
    print(f"\nConclusion: Since P-Value ({p_value:.4f}) >= Alpha ({alpha}), we FAIL TO REJECT the null hypothesis.")
    print("There is not enough evidence to say that the typical property price in Altona has increased beyond $800,000.")
```

# Task 2. For the year 2016, is there any difference in the prices of properties sold in the
 summer months vs winter months?
• Consider months from October till March as winter months and rest as
summer months.
• Use a significance level of 5%

```python
# Convert Date column
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)



```

```python
# Null Hypothesis (H0): There is no difference in the mean prices of summer vs. winter months (mu{summer} = {winter}).
# Alternative Hypothesis (H1): there is a significant difference in the mean prices (mu{summer} \mu{winter}).
# Significance Level (alpha): 0.05 (5%).


# Winter: October, November, December, January, February, March.
# Summer: April, May, June, July, August, September.






# Filter for the year 2016
df_2016 = df[df['Date'].dt.year == 2016].copy()

# Categorize into Winter vs Summer based on user definition
# Winter: Oct(10), Nov(11), Dec(12), Jan(1), Feb(2), Mar(3)
winter_months = [10, 11, 12, 1, 2, 3]

def categorize_season(month):
    if month in winter_months:
        return 'Winter'
    else:
        return 'Summer'

df_2016['Season'] = df_2016['Date'].dt.month.apply(categorize_season)

#Extract prices for both groups
summer_prices = df_2016[df_2016['Season'] == 'Summer']['Price'].dropna()
winter_prices = df_2016[df_2016['Season'] == 'Winter']['Price'].dropna()

# Descriptive Statistics
print(f"--- 2016 Price Comparison ---")
print(f"Summer (Apr-Sep) Mean: ${summer_prices.mean():,.2f} (Count: {len(summer_prices)})")
print(f"Winter (Oct-Mar) Mean: ${winter_prices.mean():,.2f} (Count: {len(winter_prices)})")

# Perform Independent Two-Sample T-Test
# We use equal_var=False (Welch's T-test) as a safer default if variances are unequal
t_stat, p_value = stats.ttest_ind(summer_prices, winter_prices, equal_var=False)

print(f"\n--- Hypothesis Test Results ---")
print(f"T-Statistic: {t_stat:.4f}")
print(f"P-Value: {p_value:.4f}")


alpha = 0.05
if p_value < alpha:
    print(f"\nConclusion: Since P-Value ({p_value:.4f}) < Alpha ({alpha}), we REJECT the null hypothesis.")
    print("There IS a statistically significant difference in property prices between summer and winter months in 2016.")
else:
    print(f"\nConclusion: Since P-Value ({p_value:.4f}) >= Alpha ({alpha}), we FAIL TO REJECT the null hypothesis.")
    print("There is NO statistically significant difference in property prices between summer and winter months in 2016.")
```

## 3. For the suburb of Abbotsford, what is the probability that out of 10 properties sold, 3 will not have a car parking space?
• Use the column car in the dataset.
• Round off your answer to 3 decimal places

```python
# Filter for Abbotsford
abbotsford_data = df[df['Suburb'] == 'Abbotsford']


# Calculate the probability 'p' of a property having NO car space (Car == 0)
# we drop NaNs in the Car column for an accurate proportion
total_properties = abbotsford_data['Car'].dropna().count()
no_car_properties = (abbotsford_data['Car'] == 0).sum()
p = no_car_properties / total_properties
print(f"--- Abbotsford Car Space Stats ---")
print(f"Total properties: {total_properties}")
print(f"Properties with 0 car spaces: {no_car_properties}")
print(f"Probability (p) of no car space: {p:.4f}")




#Binomial Distribution calculationy
# n = 10 properties, k = 3 properties with no car space
from scipy.stats import binom 
n = 10
k = 3
prob_exactly_3 = binom.pmf(k, n, p)
print(f"\n--- Binomial Probability ---")
print(f"Probability that exactly 3 out of 10 have no car space: {prob_exactly_3:.3f}")
```

# Task 6: One-Sample Hypothesis Test (Industry Pricing) A real estate firm claims that the average property price in Richmond is $1,000,000. Using the dataset, test whether the actual average price is significantly different from this claim at a 5% significance level. Clearly state:
* Null and alternative hypotheses
• Test statistic
• p-value
• Final business conclusion

```python
# filter for Richmond
richmond_prices = df[df['Suburb'] == 'Richmond']['Price'].dropna()


# Null Hypothesis (H0): The average property price in Richmond is equal to $1,000,000 (mu = 1,000,000).
# Alternative Hypothesis (H1): The average property price in Richmond is significantly different from $1,000,000 (mu = 1,000,000).
# Significance Level (alpha): 0.05 (5%)



#Descriptive Statistics
mean_price = richmond_prices.mean()
n_samples = len(richmond_prices)

print(f"--- Statistics for Richmond ---")
print(f"Number of properties: {n_samples}")
print(f"Sample Mean Price: ${mean_price:,.2f}")

# Perform Two-Tailed One-Sample T-Test
# Population mean under H0
pop_mean_h0 = 1000000

# stats.ttest_1samp performs a two-tailed test by default
t_stat, p_value = stats.ttest_1samp(richmond_prices, pop_mean_h0)

print(f"\n--- Hypothesis Test Results ---")
print(f"Test Statistic (t): {t_stat:.4f}")
print(f"P-Value: {p_value:.4f}")

# Final Business Conclusion
alpha = 0.05
print("\n--- Final Business Conclusion ---")
if p_value < alpha:
    print(f"Since the p-value ({p_value:.4f}) is less than the significance level ({alpha}), we REJECT the null hypothesis.")
    print("Conclusion: There is sufficient evidence to conclude that the actual average property price in Richmond is significantly different from $1,000,000.")
else:
    print(f"Since the p-value ({p_value:.4f}) is greater than or equal to the significance level ({alpha}), we FAIL TO REJECT the null hypothesis.")
    print("Conclusion: There is not enough evidence to claim that the average property price in Richmond is significantly different from $1,000,000. The firm's claim is plausible.")
```

## Task 7: Independent Two-Sample T-Test (Feature Impact)
Do properties with car parking sell at a higher average price than properties without
car parking, across the entire dataset? Use a 5% significance level and justify:
• Choice of test
• Interpretation of p-value
• Business implications for developers


```python
# Null Hypothesis (H0): Properties with car parking have the same or lower average price compared to properties without car parking (mu{with_car} = mu{without_car}).
# Alternative Hypothesis (H1): Properties with car parking sell at a higher average price than properties without car parking (mu{with_car} > mu{without_car}).
# Significance Level (alpha): 0.05 (5%)




#drop rows where Price or Car is missing
clean_df = df.dropna(subset=['Price', 'Car'])

#Split into two groups
with_car = clean_df[clean_df['Car'] > 0]['Price']
without_car = clean_df[clean_df['Car'] == 0]['Price']

# Descriptive Statistics
print("--- Price Statistics ---")
print(f"With Car Parking: n = {len(with_car)}, Mean Price = ${with_car.mean():,.2f}")
print(f"Without Car Parking: n = {len(without_car)}, Mean Price = ${without_car.mean():,.2f}")

# Perform Independent Two-Sample T-Test
t_stat, p_value = stats.ttest_ind(with_car, without_car, equal_var=False, alternative='greater')

print("\n--- Hypothesis Test Results ---")
print(f"Test Statistic (t): {t_stat:.4f}")
print(f"P-Value: {p_value:}")

# Conclusion
alpha = 0.05
print("\n--- Conclusion ---")
if p_value < alpha:
    print(f"Since p-value < {alpha}, we REJECT the null hypothesis.")
    print("Properties with car parking sell at a statistically significant HIGHER average price.")
else:
    print(f"Since p-value >= {alpha}, we FAIL TO REJECT the null hypothesis.")
    print("There is no significant evidence to say properties with car parking sell at a higher price.")





```

## Task: 8 Two-Way ANOVA (Location & Property Type)
Investigate whether property prices are influenced by:
• Suburb
• Type of property
• Interaction between suburb and property type
Use Two-Way ANOVA and explain which factors significantly affect price.


```python
import statsmodels.api as sm
from statsmodels.formula.api import ols

# clean missing values for relevant columns
clean_df = df.dropna(subset=['Price', 'Suburb', 'Type'])

# Build the Ordinary Least Squares (OLS) model
# The formula 'Price ~ C(Suburb) + C(Type) + C(Suburb):C(Type)' specifies the two main effects and their interaction.
# Note: This might take a few moments to run depending on the number of unique suburbs.
model = ols('Price ~ C(Suburb) + C(Type) + C(Suburb):C(Type)', data=clean_df).fit()

# Perform Two-Way ANOVA
anova_table = sm.stats.anova_lm(model, typ=2)

print("--- Two-Way ANOVA Results ---")
print(anova_table)
```

## 10. Industry-Style Hypothesis Validation (Policy Decision)
A housing policy group believes that properties with more than 2 bathrooms
command a premium price.
Design and execute a statistical test to validate this claim:
• Identify the correct test
• State hypotheses
• Report p-value
• Give a clear recommendation to policymakers


```python


#drop missing values for Price and Bathroom
clean_df = df.dropna(subset=['Price', 'Bathroom'])

#Split into the two groups
more_than_two_baths = clean_df[clean_df['Bathroom'] > 2]['Price']
two_or_less_baths = clean_df[clean_df['Bathroom'] <= 2]['Price']

#Show the basic means
print("--- Average Prices ---")
print(f"More than 2 Bathrooms: ${more_than_two_baths.mean():,.2f}")
print(f"2 or Fewer Bathrooms: ${two_or_less_baths.mean():,.2f}")

#Perform the Independent Right-Tailed T-Test
t_stat, p_value = stats.ttest_ind(more_than_two_baths, two_or_less_baths, equal_var=False, alternative='greater')

print("\n--- Statistical Test Results ---")
print(f"Test Statistic (t): {t_stat:.4f}")
print(f"P-Value: {p_value:.4e}")  # Uses scientific notation as the p-value will be very small

#Conclusion
if p_value < 0.05:
    print("\nResult: REJECT the Null Hypothesis. Properties with >2 bathrooms command a premium.")
else:
    print("\nResult: FAIL TO REJECT the Null Hypothesis.")
```
