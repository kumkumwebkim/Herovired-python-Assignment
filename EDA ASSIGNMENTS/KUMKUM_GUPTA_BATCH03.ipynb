```python

numeric_df = df.select_dtypes(include=[np.number])
correlation_matrix = numeric_df.corr()

# find the strongest correlation excluding self-correlation
upper_tri = correlation_matrix.where(np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool))
max_corr_features = upper_tri.stack().idxmax()
max_corr_value = upper_tri.stack().max()

print("Numeric columns:", numeric_df.columns.tolist())
print(f"The two features that are most strongly correlated are: {max_corr_features} with a correlation of {max_corr_value:.4f}")

plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap="coolwarm", linewidths=0.5)
plt.title("Correlation Matrix for Numeric Features")
plt.show()
```

```python
# which country negligible office supply orders?
plt.figure(figsize=(10,6))
sns.barplot(x=country_counts.index, y=country_counts.values)
plt.title('Office Supplies Orders by Country')
plt.xlabel('Country')
plt.ylabel('Count')
plt.xticks(rotation=90)
plt.show()
```

#### The countries with negligible office supply orders (defined as having the absolute lowest transaction counts of only **1 order** each) are **Armenia, Bahrain, Eritrea, and Swaziland**. ### Why these countries have negligible orders:

a. **Market Penetration and Operational Focus:** This dataset represents a multinational retail operation (commonly known in data analytics as the *Global Superstore* dataset). The company’s primary logistics, supply chains, and marketing efforts are heavily concentrated in major global economies or high-density consumer regions like the United States, France, Australia, Mexico, and Germany. Smaller or developing markets receive very little corporate focus.


b. **Geographic and Supply Chain Constraints:** Establishing shipping channels, managing customs, and setting up local distribution centers for small or landlocked nations (such as Swaziland, Lesotho, or Armenia) is often cost-prohibitive for an international retailer. Consequently, any orders appearing from these nations are typically sporadic, one-off purchases rather than established, recurring revenue streams.

c. **Low Local Brand Awareness:** Without active localized marketing, e-commerce visibility, or corporate contracts in these specific countries, consumer awareness is near zero, leading to a negligible footprint in the transaction history.
